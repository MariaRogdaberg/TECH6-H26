```python
import pandas as pd

# Use this path to use the CSV file from the data/ directory
file = 'lending_club.csv'

df = pd.read_csv(file, sep=',')

# Variables used in the analysis
variables = [
    'interest_rate',
    'loan_amount',
    'term',
    'annual_income',
    'debt_to_income',
    'emp_length',
    'homeownership',
    'verified_income',
    'application_type',
    'loan_purpose',
    'credit_history_years',
    'inquiries_last_12m',
    'open_credit_lines',
    'total_credit_limit',
    'credit_utilization',
    'delinq_2y',
    'bankruptcy',
]

n_before = len(df)

# Drop rows with any missing observation
df = df.dropna(subset=variables)

n_after = len(df)

print(f'Number of observations before: {n_before:,d}')
print(f'Dropped: {n_before - n_after:,d}')
print(f'Number of observations after: {n_after:,d}')
```
